```python
import pandas as pd
import matplotlib
```

```python
polya_df = pd.read_csv('/Users/selammehreteab/Downloads/polya_results_with_genenames.tsv', sep = "\t")
polya_df
```

```python
polya_df_pass= [polya_df['qc_tag'] == 'PASS'].copy()
polya_df_pass
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu
import numpy as np

palette = {
    "wtdmso": "gray",
    "wtcsc": "red",
    "mtdmso": "blue",
    "mtcsc": "purple"
}
order = ["wtdmso", "wtcsc", "mtdmso", "mtcsc"]

# Compute stats on FULL data (not trimmed)
ref = "wtdmso"
comparisons = ["wtcsc", "mtdmso", "mtcsc"]

def get_stars(p):
    if p < 0.0001:
        return "****"
    elif p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return "ns"

ref_vals = polya_df_pass.loc[polya_df_pass["condition"] == ref, "polya_length"].dropna()

pvals = {}
for comp in comparisons:
    comp_vals = polya_df_pass.loc[polya_df_pass["condition"] == comp, "polya_length"].dropna()
    stat, p = mannwhitneyu(ref_vals, comp_vals, alternative="two-sided")
    pvals[comp] = p

# For PLOTTING only, clip y-axis to the 1st-99th percentile range
y_low, y_high = np.percentile(polya_df_pass["polya_length"].dropna(), [1, 99])

plt.figure(figsize=(8, 6))
ax = sns.boxplot(data=polya_df_pass, x="condition", y="polya_length",
                  order=order, palette=palette, showfliers=False)

plt.xlabel("Condition")
plt.ylabel("Poly(A) length")
plt.title("Poly(A) tail length by condition")
plt.xticks(rotation=45)

# Brackets, spaced relative to the clipped range
y_range = y_high - y_low
h = y_range * 0.05
step = y_range * 0.08
x_positions = {cond: i for i, cond in enumerate(order)}
ref_x = x_positions[ref]
current_y = y_high + step

for i, comp in enumerate(comparisons):
    stars = get_stars(pvals[comp])
    comp_x = x_positions[comp]
    y = current_y + i * step
    ax.plot([ref_x, ref_x, comp_x, comp_x], [y, y + h, y + h, y], lw=1.2, c="black")
    ax.text((ref_x + comp_x) / 2, y + h, stars, ha="center", va="bottom", fontsize=12)

plt.ylim(y_low - y_range * 0.05, current_y + len(comparisons) * step + step)
plt.tight_layout()
# plt.savefig("polya_boxplot.png", dpi=300)
plt.show()
```
